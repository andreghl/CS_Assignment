```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("istherecorrelation.csv", sep = ";")

x = data["WO [x1000]"]
y = data["NL Beer consumption [x1000 hectoliter]"]
z = data["Year"]

plt.title("Is there correlation?")
plt.scatter(x, y)
plt.xticks(rotation = 90)
plt.show()
```

```python
plt.title("Is there correlation?")
plt.plot(z, y, label = "beer")
plt.plot(z, x, label = "WO")
plt.show()
```
